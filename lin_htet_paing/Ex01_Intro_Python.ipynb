# Ex 4: Daily Maths

You are working in a retail shop. Write a program that computes the total cost for a customer that bought 3 differnt types of fruits : apple, orange, mango. Ask the user to enter the quantity and price for each type.

```python
#accepting inputs from user
input_apple = input("Enter the quantity and price of apple. e.g (1,10)")
input_orange = input("Enter the quantity and price of orange. e.g (2,10)")
input_mango = input("Enter the quantity and price of mango. e.g (3,10)")

#splitting the input data into price and quantity as array
#eg apple[0] is quantity and apple[1] is price
apple = input_apple.split(",")
orange = input_orange.split(",")
mango = input_mango.split(",")

#calculating the price of each fruit
price_of_apple = int(apple[0]) * float(apple[1])
price_of_orange = int(orange[0]) * float(orange[1])
price_of_mango = int(mango[0]) * float(mango[1])

#calculating the total cost
total_price = price_of_apple + price_of_orange + price_of_mango

#showing the result to user
print("Price of Apple is {}".format(price_of_apple))
print("Price of Orange is {}".format(price_of_orange))
print("Price of Mango is {}".format(price_of_mango))
print("Total cost is {}".format(total_price))
```

# Ex 8: Computing.
Write a program that asks the user to enter two numbers, x and y, and computes |x-y|/(x+y).

```python
#getting the two number from user
xy = input("Enter two number x and y e.g x,y: ")

#splitting the two number by using split function
x, y = xy.split(",")

#parsing the splitted string into integer
x = int(x)
y = int(y)

#computing the |x-y|/(x+y) formula
result = abs(x-y)/(x+y)
print("Result is {}".format(result))
```

# Ex 9: Computing

Write a program that asks the user for a number of seconds and prints out how many minutes and seconds that is. For instance, 200 seconds is 3 minutes and 20 seconds.

```python
#getting input from user and convert into int
seconds_input = int(input("Enter the number of seconds: "))

#calculating the minutes by using bultin divmod function
minutes, seconds= divmod(seconds_input, 60)
print("{} seconds is {} minutes and {} seconds".format(seconds_input, minutes, seconds))

```

# Ex 10: Using Math Module

Write a program that asks the user for a number and then prints out the sine, cosine, and tangent of that number.

```python
import math

num = float(input("Enter a number:"))
sine_value = math.sin(num)
cosine_value = math.cos(num)
tangent_value = math.tan(num)
print(sine_value)
print(cosine_value)
print(tangent_value)
```
