Ex 3: Write a function called number_of_factors that takes an integer and returns how many
factors the number has.

```python
"""
number_of_factors calculate the count of factors of a number that tway given as a parameter
Args:
  n(integer)
Returns:
  count(integer)
"""
def number_of_factors(n):
  count = 0
  for i in range (1, n+1):
    if (n%i == 0):
      count += 1
  return count;

number_of_factors(1)
```

Ex4: Write a function called first_diff that is given two strings and returns the first location in
which the strings differ. If the strings are identical, it should return -1.Ignore the case in your function. It means h and H are the same.

```python
"""
first_diff calculate the first location in which the strings differ
Args:
  first(string)
  second(string)
Returns:
  index(integer)
"""
def first_diff(first, second):
  first = first.lower()
  second = second.lower()
  if (len(first) != len(second)):
    return -1
  for i in range(len(first)):
    if first[i] == second[i]:
      return i

  return -1

print("The first location in which strings differ is index {}".format(first_diff("aello", "Hello")))
```

Ex6: Write a function that solves the system of equations ax + by = e and cx +dy = f. Your function should takes 6 parameters, (a, b, c, d, e,f) and return the values of x and y.

```python
"""
calculate_equation calculate the values of x and y

Args:
  a(integer)
  b(integer)
  c(integer)
  d(integer)
  e(integer)
  f(integer)
Returns:
  x(integer)
  y(integer)
"""
def calculate_equation(a,b,c,d,e,f):
  dividor = a*d - b*c
  x = (e*d - b*f)/dividor
  y = (a*f - e*c)/dividor
  return x,y

x,y = calculate_equation(1,2,3,4,5,6)
print("The value of x is {} and y is {}".format(x,y))
```

Ex8:  Write a function called primes that is given a number n and returns a list of the first n  primes. Let the default value of n be 100.

```python
def is_prime(num: int) -> bool:
    """Helper function to check if a number is prime."""
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def primes(n: int = 100) -> list[int]:
    """Returns a list of the first n prime numbers."""
    prime_list = []
    candidate = 2

    while len(prime_list) < n:
        if is_prime(candidate):
            prime_list.append(candidate)
        candidate += 1

    return prime_list

print(primes(10))
```

Ex9: A Tic-tac-toe board can be represented be a 3×3 two-dimensional list, where zeroes stand for  empty cells, ones stand for X’s and twos stand for O’s. Write a function that is given such a list and randomly chooses a spot in which to place  a 2 (O). The spot chosen must currently be empty and a spot must be chosen.

```python
import random

def place_random_o(board):
    # Collect all empty positions (cells with 0)
    empty_positions = []

    for row in range(3):
        for col in range(3):
            if board[row][col] == 0:
                empty_positions.append((row, col))

    # If there are no empty spots, raise an error
    if not empty_positions:
        raise ValueError("No empty spots available")

    # Randomly choose one empty position
    row, col = random.choice(empty_positions)

    # Place O (2) in the chosen spot
    board[row][col] = 2

    return board

board = [
    [1, 0, 2],
    [0, 1, 0],
    [2, 0, 0]
]

updated_board = place_random_o(board)

for row in updated_board:
    print(row)
```

Ex10:  Write a function that is given a 9 × 9 potentially solved Sudoku and returns True if it is solved correctly and False if there is a mistake. The Sudoku is correctly solved if there are no repeated numbers in any row or any column or in any of the nine “blocks.”


```python
def is_solved_sudoku(board: list[list[int]]) -> bool:
    """
    Validates a 9x9 Sudoku board.
    Returns True if solved correctly, False otherwise.
    """

    def is_valid_unit(unit: list[int]) -> bool:
        # A valid unit must contain exactly 1-9 with no duplicates
        return sorted(unit) == list(range(1, 10))

    # 1. Check Rows
    for row in board:
        if not is_valid_unit(row):
            return False

    # 2. Check Columns
    for col in range(9):
        column_values = [board[row][col] for row in range(9)]
        if not is_valid_unit(column_values):
            return False

    # 3. Check 3x3 Blocks
    # Iterate through the top-left corner of each 3x3 block
    for r in range(0, 9, 3):
        for c in range(0, 9, 3):
            block = []
            for i in range(3):
                for j in range(3):
                    block.append(board[r + i][c + j])

            if not is_valid_unit(block):
                return False

    return True

# A correctly solved Sudoku board
valid_board = [
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
]

# An invalid board (first two numbers swapped in the first row)
invalid_board = [
    [3, 5, 4, 6, 7, 8, 9, 1, 2], # Errors here
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
]

print(f"Is the valid board solved? {is_solved_sudoku(valid_board)}")
print(f"Is the invalid board solved? {is_solved_sudoku(invalid_board)}")
```
